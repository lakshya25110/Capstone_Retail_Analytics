```python

spark.conf.set(
    "fs.azure.account.key.laksstorage25.dfs.core.windows.net",
    dbutils.secrets.get(scope="dbscope25", key="kyy-secret")
)
```

```python
sales_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("abfss://bronze@laksstorage25.dfs.core.windows.net/raw_data/sales_transactions.csv")
```

```python
sales_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("capstone_catalog_demo.bronze.sales")
```
